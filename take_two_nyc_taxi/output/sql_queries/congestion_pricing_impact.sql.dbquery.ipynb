```sql

-- Business Question 3:
-- Compare inbound Manhattan CBD taxi trips before and after
-- congestion pricing implementation on January 5, 2025.

SELECT
    period,
    service_type,
    trip_count,
    paid_trip_count,
    ROUND(collected_revenue, 2) AS collected_revenue,
    average_fare,
    ROUND(total_cbd_congestion_fee, 2) AS total_cbd_congestion_fee,
    trips_with_cbd_fee,
    period_days,
    average_trips_per_day,
    average_revenue_per_day
FROM workspace.nyc_taxi_gold.congestion_pricing_impact
ORDER BY period, service_type;
```

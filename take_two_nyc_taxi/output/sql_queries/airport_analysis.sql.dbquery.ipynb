```sql

-- Business Question 2:
-- Airport trip volume, average fare and tipping behaviour
-- by month, airport, service type and direction.

SELECT
    month,
    airport,
    service_type,
    direction,
    trip_count,
    ROUND(average_fare, 2) AS average_fare,
    credit_card_trip_count,
    observed_card_tip_count,
    tipped_card_trip_count,
    ROUND(average_card_tip, 2) AS average_card_tip,
    ROUND(average_positive_card_tip, 2) AS average_positive_card_tip,
    card_tip_rate_pct
FROM workspace.nyc_taxi_gold.airport_analysis
ORDER BY month, airport, direction, service_type;
```

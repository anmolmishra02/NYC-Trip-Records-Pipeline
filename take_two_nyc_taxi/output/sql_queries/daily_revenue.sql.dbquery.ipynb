```sql

-- Business Question 1:
-- Daily revenue by pickup borough and service type.

SELECT
    pickup_date,
    pickup_borough,
    service_type,
    trip_count,
    collected_trip_count,
    ROUND(collected_revenue, 2) AS collected_revenue,
    excluded_payment_trip_count
FROM workspace.nyc_taxi_gold.daily_revenue
ORDER BY pickup_date, pickup_borough, service_type;
```
